# Mapping exercise - Hochgrat
![map example](example_map.png)
Aim of this notebook is to showcase a complete modelling workflow, involving field measurements, data processing and remote sensing data. Study area is the Hochgrat region in Allgäu, Germany. In a fieldtrip, students gathered orientation data of structural features and bedding of molasse sediments. Remote sensing data from BayernAtlas.de was utilized in the form of a Digital Terrain Model (DTM) with a 1mx1m resolution. In QGIS, the corresponding slope was computed. Raster files and shapefiles are displayed in a leafmap map. To run this notebook, we suggest starting with a new python environment, as leafmap can be tricky to install, according to its developers. For the development of this notebook, Mamba was used and the following installation steps were conducted:

# Preprocessing orientation data
The data collected by student contains multiple measurements per location and variable kinds of observations. Observations include faults (B/Störungen, B/S), Bedding(B), and Fractures (F). Therefore, we sample only for B and display meausrements per location. As there are no large outliers, we also average over all meausrements per location, using <code>mplstereonet.kmeans</code>. In a next step, the average location in X, Y is utilized to sample the elevation from the DTM. This combined data of kmeans average dip direction/dip and position can be stored and utilized in the modelling workflow.

> **_NOTE_**: Different coordinate systems are a common problem in geo modelling. Make sure to use the same coordinate system (UTM 32N in this case) in all modelling steps.

````python
ori_gdf["type"].hist()
````

````python
dtm = rasterio.open("Hochgrat_dtm_1x1.tif")
print(dtm.crs)
dtm.close()
````

````python
ori_gdf.crs
````

````python
ori_gdf = ori_gdf.to_crs("25832")
ori_gdf[["x", "y"]] = ori_gdf.geometry.get_coordinates()
````

````python
beddings_gdf = ori_gdf.loc[ori_gdf["type"] == "B"]

locations = beddings_gdf["Location #"].unique()

# create storage dataframe for average orientations per location in gempy shape
ori_df = pd.DataFrame([], columns=["X", "Y", "Z", "formation", "azimuth", "dip", "polarity"])

i = 0
for iloc in locations:
    subset = beddings_gdf.loc[beddings_gdf["Location #"] == iloc]

    fig = plt.figure()
    ax = plt.subplot(projection='stereonet')
    ax.plane(subset["azimuth"]+90, subset["dip"], c='k', alpha=0.5)
    ax.pole(subset["azimuth"]+90, subset["dip"], c='k', alpha=0.5)
    mean_dip = mplstereonet.kmeans(subset["azimuth"]+90, subset["dip"],num=1)
    strike_cent, dip_cent = mplstereonet.geographic2pole(*zip(*mean_dip))
    ax.pole(strike_cent, dip_cent, marker="^", color="red")
    ax.grid()
    fig.suptitle("Location %i (n = %i) $\mu_{azimuth}$ %.1f, $\mu_{dip}$ %.f"%(iloc, len(subset), strike_cent-90, dip_cent))

    # write data to dataframe. Note: elevation is missing and will be sampled in a later step
    ori_df.loc[i] = [subset["x"].mean(), subset["y"].mean(), 9999, "Location %i"%iloc,  strike_cent[0]-90, dip_cent[0], 1]
    i+=1
    plt.show()

dtm.close()
````

# Use rasterio to sample elevation from DTM

````python
coords = ori_df[["X", "Y"]].values

dtm = rasterio.open("Hochgrat_dtm_1x1.tif")
elevations = [z[0] for z in dtm.sample(coords)]
dtm.close()

ori_df["Z"] = elevations

ori_df
````

# Display orientation data and mapped interfaces to surfaces for geomodel (should be replaced with leafmap in future)

````python
surf_gdf = gpd.read_file("shp/molasse_interface_lines.shp")

fig, ax = plt.subplots()

cmap = plt.colormaps.get_cmap("viridis")
for i in surf_gdf.index:
    surf_gdf.loc[[i]].plot(ax=ax, label=surf_gdf.loc[i, "formation"] + " i=" + str(i), color=cmap((i+1)/len(surf_gdf)))
    
for i in ori_df.index:
    plt.text(x=ori_df.loc[i, "X"], y=ori_df.loc[i,"Y"]+(i%2)*50, s=ori_df.loc[i, "formation"]+ " i=" + str(i)) # i%2 becase location 26 and 27 were overlaying each other, so every uneven index is shifted by 50

plt.legend()
plt.axis("equal");
````

````python
xmin = surf_gdf.get_coordinates()["x"].min(axis=0)
ymin = surf_gdf.get_coordinates()["y"].min(axis=0)
xmax = surf_gdf.get_coordinates()["x"].max(axis=0)
ymax = surf_gdf.get_coordinates()["y"].max(axis=0)
````

````python
geo_model: gp.data.GeoModel = gp.create_geomodel(
    project_name='Hochgrat Mapping',
    extent=[xmin, xmax, ymin, ymax, 1000, 2000],
    resolution=[100, 100, 100],
    refinement=3,
    structural_frame=gp.data.StructuralFrame.initialize_default_structure()
)
````

````python
geo_model.structural_frame
````

````python
# add first layer
# we assign lineament 3 and 5 to parts of one surface. 
# furthermore, orientation measurement 16, 26 and 27 (idx 3, 9, 10) belong to this layer as well


# add surface points
lineament_selection = surf_gdf.loc[[0,2,4]]

# extract x and y coordinates
x_sp = lineament_selection.get_coordinates().values[:,0]
y_sp = lineament_selection.get_coordinates().values[:,1]

# sample elevation from DTM
dtm = rasterio.open("Hochgrat_dtm_1x1.tif")
z_sp = np.array([z[0] for z in dtm.sample(lineament_selection.get_coordinates().values)])
dtm.close()

gp.add_surface_points(
    geo_model=geo_model,
    x=x_sp,
    y=y_sp,
    z=z_sp,
    elements_names=['surface1']*len(x_sp)
    )

# add orientations
ori_selections = ori_df.loc[[0,3,5,6,9,10,8]]
x_or = ori_selections["X"].values
y_or = ori_selections["Y"].values
z_or = ori_selections["Z"].values
orientation = ori_selections[["azimuth", "dip", "polarity"]].values

for i in range(len(x_or)):
    gradient = convert_orientation_to_pole_vector([orientation[i,0]], [orientation[i,1]], [orientation[i,2]])
    gp.add_orientations(
        geo_model=geo_model,
        x=[x_or[i]],
        y=[y_or[i]],
        z=[z_or[i]],
        pole_vector=gradient,
        elements_names="surface1"
    )
    
````

````python
geo_model.update_transform(gp.data.GlobalAnisotropy.NONE)
gp.compute_model(geo_model, engine_config=gp.data.GemPyEngineConfig())
````

````python
gpv.plot_2d(geo_model, direction="x")
````

````python
gpv.plot_3d(geo_model)
````

````python
# add second layer
# we assign lineament 2 to the oldest surface. 
# furthermore, orientation measurement 13 and 15 (idx 1, 2) belong to this layer as well


# add surface points
lineament_selection = surf_gdf.loc[[2]]

# extract x and y coordinates
x_sp = lineament_selection.get_coordinates().values[:,0]
y_sp = lineament_selection.get_coordinates().values[:,1]

# sample elevation from DTM
dtm = rasterio.open("Hochgrat_dtm_1x1.tif")
z_sp = np.array([z[0] for z in dtm.sample(lineament_selection.get_coordinates().values)])
dtm.close()

element2 = gp.data.StructuralElement(
    name='surface2',
    color=next(geo_model.structural_frame.color_generator),
    surface_points=gp.data.SurfacePointsTable.initialize_empty(),
    orientations=gp.data.OrientationsTable.initialize_empty()
)

geo_model.structural_frame.structural_groups[0].append_element(element2)

gp.add_surface_points(
    geo_model=geo_model,
    x=x_sp,
    y=y_sp,
    z=z_sp,
    elements_names=['surface2']*len(x_sp)
    )

# add orientations
ori_selections = ori_df.loc[[1,2]]
x_or = ori_selections["X"].values
y_or = ori_selections["Y"].values
z_or = ori_selections["Z"].values
orientation = ori_selections[["azimuth", "dip", "polarity"]].values

for i in range(len(x_or)):
    gradient = convert_orientation_to_pole_vector([orientation[i,0]], [orientation[i,1]], [orientation[i,2]])
    gp.add_orientations(
        geo_model=geo_model,
        x=[x_or[i]],
        y=[y_or[i]],
        z=[z_or[i]],
        pole_vector=gradient,
        elements_names="surface2"
    )
    
````

````python
# add thrid layer
# we assign lineament 4 to the youngest surface. 
# furthermore, orientation measurement 24 (idx 7) belongs to this layer as well


# add surface points
lineament_selection = surf_gdf.loc[[3]]

# extract x and y coordinates
x_sp = lineament_selection.get_coordinates().values[:,0]
y_sp = lineament_selection.get_coordinates().values[:,1]

# sample elevation from DTM
dtm = rasterio.open("Hochgrat_dtm_1x1.tif")
z_sp = np.array([z[0] for z in dtm.sample(lineament_selection.get_coordinates().values)])
dtm.close()

element3 = gp.data.StructuralElement(
    name='surface3',
    color=next(geo_model.structural_frame.color_generator),
    surface_points=gp.data.SurfacePointsTable.initialize_empty(),
    orientations=gp.data.OrientationsTable.initialize_empty()
)

geo_model.structural_frame.structural_groups[0].append_element(element3)

gp.add_surface_points(
    geo_model=geo_model,
    x=x_sp,
    y=y_sp,
    z=z_sp,
    elements_names=['surface3']*len(x_sp)
    )

# add orientations
ori_selections = ori_df.loc[[1,2]]
x_or = ori_selections["X"].values
y_or = ori_selections["Y"].values
z_or = ori_selections["Z"].values
orientation = ori_selections[["azimuth", "dip", "polarity"]].values

for i in range(len(x_or)):
    gradient = convert_orientation_to_pole_vector([orientation[i,0]], [orientation[i,1]], [orientation[i,2]])
    gp.add_orientations(
        geo_model=geo_model,
        x=[x_or[i]],
        y=[y_or[i]],
        z=[z_or[i]],
        pole_vector=gradient,
        elements_names="surface3"
    )
    
````

# Create georeferenced geo map for GIS application
To complete the workflow, we finalize the modelling by creating a geological map for the model area. 

````python
XX, YY = np.meshgrid(np.linspace(xmin, xmax, 200), np.linspace(ymin, ymax, 200))
map_coords = np.hstack([XX.reshape(-1,1), YY.reshape(-1,1)])

# sample elevation from DTM
dtm = rasterio.open("Hochgrat_dtm_1x1.tif")
z_map = np.array([z[0] for z in dtm.sample(map_coords)])
dtm.close()

map_xyz = np.hstack([map_coords, z_map.reshape(-1,1)])
````

````python
geo_sol : np.ndarray = gp.compute_model_at(
    gempy_model=geo_model,
    at=map_xyz
)
geo_sol
````

This can already be displayed in matplotlib, for example like this:

````python
geo_map = np.flipud(geo_sol.reshape(XX.shape))
plt.imshow(geo_map, extent=[xmin, xmax, ymin, ymax], cmap="Set1")
````

Ideally, however, we will want to display the map in GIS software, so we have to generate a georeferenced tif file. 

````python
# Create transform matrix to write tif
transform = rasterio.transform.from_bounds(xmin, ymin, xmax, ymax, XX.shape[0], XX.shape[1])

# write tif
with rasterio.open(
        "geomap_Hochgrat.tif",
        mode="w",
        driver="GTiff",
        height=geo_map.shape[0],
        width=geo_map.shape[1],
        count=1,
        dtype=geo_map.dtype,
        crs=dtm.crs,
        transform=transform,
) as new_dataset:
        new_dataset.write(geo_map, 1)
````
